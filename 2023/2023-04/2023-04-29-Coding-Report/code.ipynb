## Problem Description

This report demonstrates a method for solving linear matrix equation. It covers general non-pivoted Gaussion Delimination Method and LU Decomposition Method. Given a linear equation

$$ Ax = b,\qquad A\in\mathbb{A}^{n\times n},b\in\mathbb{R}^n $$

**Algorithm**

Adopting **Gaussion Delimination Method**, we apply row transform sequentially, until the coefficient matrix $A$ becomes upper triangular,

$$ R_{m}R_{m-1}\cdots R_1 Ax = Ux = R_{m}R_{m-1}\cdots R_1 b = y $$

Then we use backword substitution algorithm to calculate $x$.

Adopting **LU Decomposition Method**, we first decomposite $A$ into the multiplication of a lower triangular matrix and an upper triangular matrix. 

$$ Ax=LUx=b $$

Letting $Ux=y$, then $Ly=b$, we can then apply backward substitution algorithm to solve for $y$, and next apply forward substitution algorithm to solve for $x$ in $Ux=y$.

In fact, $L=R_{m}R_{m-1}\cdots R_1$, therefore the two methods are essentially the same.

The **backward substitution algorithm** performed on an upper triangular $n\times n$ matrix:

$$
U=\begin{bmatrix} a_{11} & a_{12} & a_{13} & \cdots & a_{1n}\\
0 & a_{22} & a_{23} & \cdots & a_{2n}\\
\vdots & \vdots & \ddots & \vdots\\
0&0 & 0&\cdots & a_{nn}\\
\end{bmatrix} $$

Note that the last row contains only one non-zero element, hence 

$$x_{n} = b_{n} / a_{nn}$$

Next, consider the 2nd last row, it contains 2 elements, and $x_n$ is already know, hence

$$x_{n-1} = (b_{n-1} - a_{n-1,n}x_{n}) / a_{n-1,n-1}$$

Repeat this operation, from the last row to the first row, we can find $x_i$ one by one. So here's the algorithm

```
for i = n, n - 1, ..., 1 do
    x(i) = b(i)
    for j = i + 1, i + 2, ..., n do
        x(i) = x(i) - u(i, j) x(j)
    end
    x(i) = x(i) / u(i, i)
end
```

Similarly, for a lower triangular matrix, we can resolve from the first row to solve $Lx=b$. This is called **forward substitution method**.

```
for i = 1, 2, ..., n do
    x(i) = b(i)
    for j = 1, 2, ..., i - 1 do
        x(i) = x(i) - u(i, j) x(j)
    end
    x(i) = x(i) / u(i, i)
end
```

**Gaussian elimination and LU Decomposition**

Gaussian's idea is to eliminate non-zero elements from the left top side of the matrix down to the right bottom side, creating an upper triangular matrix. It's done, for example on the first column, by eliminating the $1$st row from the $i$ -th row with a proper factor.  

```
for j = 1,2,...,n - 1 do
    for i = j + 1, j + 2, ..., n do
        m(i, j) = a(i, j) / a(j, j)
        for k = j + 1, j + 2, ..., n do
            a(i, k) = a(i, k) - m(i, j) a(j, k)
        end
        b(i) = b(i) - m(i, j) b(j)
    end
end
```

If we record the factor each time we do the elimination, then the row manipulation matrix will create a lower triangular matrix (since we always substract rows of lower index from rows of higer index). Let

$$
U=\begin{bmatrix} u_{11} & u_{12} & u_{13} & \cdots & u_{1n}\\
0 & u_{22} & u_{23} & \cdots & u_{2n}\\
\vdots & \vdots & \ddots & \vdots\\
0&0 & 0&\cdots & u_{nn}\\
\end{bmatrix}, \qquad L=\begin{bmatrix} l_{11} & 0 & 0 & \cdots & 0\\
l_{12} & l_{22} & 0 & \cdots & l_{2n}\\
\vdots & \vdots & \ddots & \vdots\\
l_{n1}&l_{n2} & l_{n3}&\cdots & l_{nn}\\
\end{bmatrix} $$

Compare coefficients, we have

$$
    a_{ik}=l_{ik}u_{kk} +\sum_{r=1}^{k-1}l_{ir}u_{rk}
$$
$$
    l_{ik}=\dfrac{\displaystyle a_{ik} - \sum_{r=1}^{k-1}l_{ir}u_{rk}}{u_{kk}}, i>k
$$

hence the algorithm

```
for j = k to n 
    u(k, j) = a(k, j) - sum([l(k, r)u(r, j) for r from 1 to k - 1])
    for i = k + 1 to n 
        l(i, k) = (a(i, k) - sum([l(i, r)u(r, k) for r from 1 to k - 1])) / u(k, k)
    end
end
```

**Time complexity**

Number of operations in elimination:

$$\sum_{i=1}^{n-1}(n-i)(2n-2i+3)=\dfrac{1}{6}(4n^3+3n^2-7n)=\dfrac{2}{3}n^3+O(n^2)=O(n^3)$$

Back & Forward substitution:

$$\dfrac{n(n-1)}{2}*2=n^2-n=O(n^2)$$

The cost of LU decomposition is the same as doing elimination.

Hence the first method has a total time complexity of $O(\dfrac{2}{3}n^3)$, so does the second method.

````python
import numpy as np
from matplotlib import pyplot as plt
from ttictoc import tic, toc
from tqdm import tqdm
from tabulate import tabulate

eps = np.finfo(float).eps
````

````python
def gaussian_elimination(A: np.ndarray, b: np.ndarray):
    n = len(A)
    U = A.copy()
    bp = b.copy()

    for j in range(0, n - 1):
        for i in range(j + 1, n):
            if abs(A[j][j]) < eps:
                return None # without pivoting, we cannot handle this
            m = U[i][j] / U[j][j]
            U[i, :] = U[i, :] - m * U[j, :]
            bp[i] -= m * bp[j]
    
    return U, bp
````

````python
def back_substitution(U: np.ndarray, b: np.ndarray) -> np.ndarray | None:
    x = np.zeros_like(b)

    if abs(U[-1, -1]) < eps:
        return None
    
    x[-1] = b[-1] / U[-1, -1]
    for i in range(len(b), 0, -1):
        x[i - 1] = (b[i - 1] - np.dot(U[i - 1, i:], x[i:])) / U[i - 1, i - 1]

    assert(np.linalg.norm(U @ x - b, 2) < 0.0001)

    return x
````

````python
def solveGE(A: np.ndarray, b: np.ndarray) -> np.ndarray | None:
    U, bp = gaussian_elimination(A, b)

    assert(np.linalg.norm(np.linalg.solve(A, b) - np.linalg.solve(U, bp), 2) < 0.0001)

    x = back_substitution(U, bp)

    assert(np.linalg.norm(A @ x - b, 2) < 0.0001)
    return x
````

````python
def forward_substitution(L: np.ndarray, b: np.ndarray) -> np.ndarray | None:
    x = np.zeros_like(b)

    if abs(L[0, 0]) < eps:
        return None
    
    x[0] = b[0] / L[0, 0]
    for i in range(1, len(b)):
        x[i] = (b[i] - np.dot(x[:i], L[i,:i])) / L[i][i]

    assert(np.linalg.norm(L @ x - b, 2) < 0.0001)
    return x
````

````python
def LUdecomposition(A: np.ndarray) -> np.ndarray:
    n = len(A)
    L = np.identity(n)
    U = A.copy()

    for j in range(0, n - 1):
        for i in range(j + 1, n):
            if abs(U[j][j]) < eps:
                return None # without pivoting, we cannot handle this
            m = U[i][j] / U[j][j]
            U[i, :] = U[i, :] - m * U[j, :]
            L[i, j] = m
    
    assert(np.linalg.norm(L @ U - A, 2) < 0.0001)
    return L, U
````

````python
def solveLUb(L: np.ndarray, U: np.ndarray, b: np.ndarray) -> np.ndarray | None:
    y = forward_substitution(L, b)
    if y is None:
        return y
    
    x = back_substitution(U, y)

    # print("y = ", y)
    # print("x = ", x)

    return x
````

````python
def solveLU(A: np.ndarray, b: np.ndarray) -> np.ndarray | None:
    L, U = LUdecomposition(A)
    # print("L", L)
    # print("U", U)
    x = solveLUb(L, U, b)
    return x
````

````python
# A = np.random.randint(20, size=(100, 100)).astype('float64') +  5 * np.sqrt(100) * np.identity(100)
A = np.array([[2, 4, 5], [7, 6, 5], [9, 11, 3]], dtype=np.float64)
print("A = ", A)

# b = np.random.randint(20, size=100).astype('float64')
b = np.array([3., 2., 1.])
print("b = ", b)

sol = np.linalg.solve(A, b)
print("reference solution x = ", sol)

x = solveGE(A, b)
print("GE x = ", x)

err = np.linalg.norm(A @ x - b, 2)
print("error = ", err)

x = solveLU(A, b)
print("LU x = ", x)

err = np.linalg.norm(A @ x - b, 2)
print("error = ", err)
````

````python
sigma = 1
mu = 0

n = 50

IR = np.random.normal(mu, sigma, size=(n, n));
````

````python
plt.hist(IR.flatten(), bins=20)
plt.xlabel("Value Intervals")
plt.ylabel("Counts")
plt.title("Matrix element distribution")
plt.savefig("hist.png")
````

````python
ns = [50, 100, 250, 500]
data = []

N = 100

for n in ns:
    print("n = ", n)
    data.append([n])

    IR = np.random.normal(mu, sigma, size=(n, n)) + 5 * np.sqrt(n) * np.identity(n)
    b = np.random.normal(mu, sigma, size=n)

    tic()
    for i in tqdm(range(N)):
        x = solveGE(IR, b)
    timeGE = toc()
    x = solveGE(IR, b)
    errGE = np.linalg.norm(IR @ x - b, 2)

    data[-1].append(timeGE)
    data[-1].append(errGE)    

    tic()
    for i in tqdm(range(N)):
        x = solveLU(IR, b)
    timeLU = toc()
    x = solveLU(IR, b)
    errLU = np.linalg.norm(IR @ x - b, 2)
    data[-1].append(timeLU)
    data[-1].append(errLU)
````

````python
print(tabulate(data, headers=["n", "Time GE", "Error GE (2-Norm)", "Time LU", "Error LU (2-Norm)"]))
````
