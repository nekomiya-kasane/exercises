```julia
function GenerateCoeffMatrix5Point(N::Int64, h::Float64, allocator::Function)
    nonzero_size = (N - 4) * (N - 4) * 5 + (N - 4) * 4 * 4 + 4 * 3 + N * 4 - 4
    ids_row = zeros(Int64, nonzero_size);
    ids_col = zeros(Int64, nonzero_size);

    values = zeros(Float64, nonzero_size);

    row = 1;
    count = 1;
    for i = 1 : N, j = 1 : N
        if i != 1 && i != N && j != 1 && j != N
            if i - 1 > 1 # up
                ids_row[count] = row;
                ids_col[count] = allocator(i - 1, j);
                values[count] = 1 / h ^ 2;
                count += 1;
            end
            if i + 1 < N # down
                ids_row[count] = row;
                ids_col[count] = allocator(i + 1, j);
                values[count] = 1 / h ^ 2;
                count += 1;
            end
            if j - 1 > 1 # left
                ids_row[count] = row;
                ids_col[count] = allocator(i, j - 1);
                values[count] = 1 / h ^ 2;
                count += 1;
            end
            if j + 1 < N # right
                ids_row[count] = row;
                ids_col[count] = allocator(i, j + 1);
                values[count] = 1 / h ^ 2;
                count += 1;
            end
        end

        # center
        ids_row[count] = row;
        ids_col[count] = allocator(i, j);
        values[count] = -4 / h ^ 2; 
        count += 1;

        # next equation
        row += 1;
    end

    return sparse(ids_row, ids_col, values);
end
```

```julia
N = 20;
L = H = 1.0;
h = L / N;

A = GenerateCoeffMatrix5Point(N, h)

# fill right
f = (x, y) -> sin(pi * x) * sin(pi * y)
f_ij = (i, j) -> f((i - 1) * h, (j - 1) * h)

g = (x, y) -> 0
g_ij = (i, j) -> g((i - 1) * h, (j - 1) * h)

id = (i, j) -> (i - 1) * N + j

rhs = zeros(Float64, size(A)[1]);

for k = 1 : N
    rhs[id(k, 1)] = -4 * g_ij(k, 1) / h ^ 2;
    rhs[id(k, N)] = -4 * g_ij(k, N) / h ^ 2;
    rhs[id(1, k)] = -4 * g_ij(1, k) / h ^ 2;
    rhs[id(N, k)] = -4 * g_ij(N, k) / h ^ 2;
end

for i = 2 : N - 1, j = 2 : N - 1
    rhs[id(i, j)] = f_ij(i, j);
end
```

```julia
@time sol = A \ rhs;
```
