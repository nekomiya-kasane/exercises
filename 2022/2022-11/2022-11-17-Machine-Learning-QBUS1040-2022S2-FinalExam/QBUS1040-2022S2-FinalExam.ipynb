```python
import numpy as np

volume_train = np.array([10.3, 10.2, 16.4, 18.8, 19.7, 15.6, 22.6, 19.9, 21. , 21.3, 19.1, 22.2, 33.8, 25.7, 24.9, 34.5, 31.7, 38.3, 42.6, 55.7, 58.3, 51.5])
volume_test = np.array([10.3, 18.2, 24.2, 21.4, 27.4, 36.3, 55.4, 51. , 77. ])

girth_train = np.array([ 8.3,  8.8, 10.5, 10.7, 10.8, 11. , 11.1, 11.2, 11.4, 11.7, 12. , 12.9, 12.9, 13.7, 13.8, 14. , 14.2, 16. , 16.3, 17.5, 17.9, 18. ])
girth_test = np.array([ 8.6, 11. , 11.3, 11.4, 13.3, 14.5, 17.3, 18. , 20.6])

height_train = np.array([70., 63., 72., 81., 83., 66., 80., 75., 76., 69., 75., 74., 85., 71., 64., 78., 80., 72., 77., 82., 80., 80.])
height_test = np.array([65., 75., 79., 76., 86., 74., 81., 80., 87.])
```

(8.1) Basis function $B_1(X)=1$, $B_2(X)=x_2$

(8.2) $ {RMS = \sqrt {\frac{1}{n}({x_1} ^ 2 + {x_2} ^ 2 + … + {x_n} ^ 2)}} $

```python
N = len(girth_train)
n = len(girth_test)

S = np.concatenate((girth_train, height_train), axis=0).reshape(2, N).T
S_test = np.concatenate((girth_test, height_test), axis=0).reshape(2, n).T

# calculate coefficients
A = np.concatenate((np.ones((N, 1)), S[:, 0:1]), axis=1)

ATA = A.T @ A
Ab = A.T @ volume_train

x = np.linalg.solve(ATA, Ab)
print("Coefficients: b1 = {:.3f}, b2 = {:.3f}".format(x[0], x[1]))

# test
A_test = np.concatenate((np.ones((n, 1)), S_test[:, 0:1]), axis=1)
b_test = (A_test @ x)

# RMS
RMS1 = np.sqrt(np.linalg.norm(b_test - volume_test) ** 2 / n)
print("RMS: {:.3f}".format(RMS1))
```

```python
A2 = np.concatenate((S, S[:, 0:1] * S[:, 1:2]), axis=1)

# calculate coefficients
ATA2 = A2.T @ A2
Ab2 = A2.T @ volume_train

x2 = np.linalg.solve(ATA2, Ab2)
print("Coefficients: b1 = {:.3f}, b2 = {:.3f}, b3 = {:.3f}".format(x2[0], x2[1], x2[2]))

# test
A_test2 = np.concatenate((S_test, S_test[:, 0:1] * S_test[:, 1:2]), axis=1)
b_test2 = (A_test2 @ x2)

# RMS
RMS2 = np.sqrt(np.linalg.norm(b_test2 - volume_test) ** 2 / n)
print("RMS: {:.3f}".format(RMS2))
```

(8.4) The second model is better due to the following reasons:

- The volumn of a tree is concerned with both its girth and height, hence we should include both of these two factors (features) into our model (this enrichs information taken into account)
- The second model has more undetermined coefficients, thus has more training freedom, and in this specific problem, 3 coefficient only will not cause over-fitting.
- According to the RMS computed above, the second model produces smaller RMS.
