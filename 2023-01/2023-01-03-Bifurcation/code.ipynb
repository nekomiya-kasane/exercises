```
# Import libraries
import numpy as np
import matplotlib.pyplot as plt
```

### A.1 Phase Portrait

```
w = 3
Y, X = np.mgrid[-w:w:100j, -w:w:100j]
U = X + X * Y
V = -Y + Y ** 2 + X * Y - X ** 3
speed = np.sqrt(U ** 2 + V ** 2)
 
fig = plt.figure(figsize = (8, 8))
plt.streamplot(X, Y, U, V, density = 2)
plt.scatter(
    np.array([0, 0, 1]),
    np.array([1, 0, -1]),
    color="orange",
    s=100
)
 
plt.show()
```

### A.2 Phase Portrait 

```
w = 4
Y, X = np.mgrid[-w:w:100j, -w:w:100j]
U = -X ** 3 + Y
V = X - Y ** 3
speed = np.sqrt(U ** 2 + V ** 2)
 
fig = plt.figure(figsize = (8, 8))
plt.streamplot(X, Y, U, V, density = 2)
plt.scatter(
    np.array([1, 0, -1]),
    np.array([1, 0, -1]),
    color="orange",
    s=100
)
 
plt.show()
```

### A.3 Phase Portrait

```
w = 4
Y, X = np.mgrid[-w:w:100j, -w:w:100j]
U = Y
V = -X + Y * (1 - 3 * X ** 2 - 2 * Y ** 2)
speed = np.sqrt(U ** 2 + V ** 2)
 
fig = plt.figure(figsize = (8, 8))
plt.streamplot(X, Y, U, V, density = 2)
plt.scatter(
    np.array([0]),
    np.array([0]),
    color="orange",
    s=100
)

rad = (1 + np.sqrt(3)) / 4;
t = np.linspace(0, 2 * np.pi, 100);
plt.plot(rad * np.cos(t), rad * np.sin(t), color='purple')
 
plt.show()
```

### A.2.(b)
