<h3>Variables and Type</h3>
<p>In python, you don't need to explicitly define the variable type. You just have to assign the value and python implictly defines the type. You can use the type(Variable_Name) to know the type of variable.<br/> A variable can hold integer, float, string, boolean etc values.</p>
<h3>Variable name rules</h3>
<p>Must be alpha-numeric i.e. can only contain alphabets and numbers. The only special character allowed is _(underscore) <br/> Must always start with an alphabet <br/> It is case sensitive i.e. "glass" and "Glass" are considered two seperate variables</p>
<h3>Operators, Comments</h3>
<p> Operators: +(addition), -(subtraction), *(multiplication), /(division), %(modulus), **(exponent), //(integer division) <br/> Single line comments: #Statements</p> 

```python
integer_variable = 10
float_variable = 12.3
string_variable = "Harry Potter"
boolean_variable = False
print("Type of",integer_variable,"is ",type(integer_variable))
print("Type of",float_variable,"is ",type(float_variable))
print("Type of",string_variable,"is ",type(string_variable))
print("Type of",boolean_variable,"is ",type(boolean_variable))

#This is an example of a single line comment
print("\nExamle of Operators: ")
x = 10
y = 3

print("x + y= ",x+y)
print("x - y= ",x-y)
print("x * y= ",x*y)
print("x / y= ",x/y)
print("x % y= ",x%y)
print("x ** y= ",x**y)
print("x // y= ",x//y)
```

<h3>Conditional operators</h3>
<p>== Equal<br/>
!= Not Equal<br/> >= Greater than or Equal<br/>
&lt= Less than or Equal<br/>
> Greater than<br/>
&lt Less than <br/>These operators are used to compare two quantities and returns the value True or False. </p>
<h3>Using 'and' & 'or'</h3>
<p>When you want to check two or more conditional statements, 'and' and 'or' keywords are to be used.<br/>
'and' is equivalent to && <br/>
'or' is equivalent to ||<br/>
Rules are as below:<br/>
Given two conditions: A, B<br/>
A and B returns True when both the statements are true<br/>
A or B returns True when either A or B is true</p>

```python
print("Checking equality of two variables:")
x = 2
y = 3
z = 2
print("x == y:",x==y)
print("x != y:",x!=y)
print("x <= y:",x<=y)
print("x >= y:",x>=y)
print("x < y:",x<y)
print("x > y:",x>y)

print("\nUing and & or")
A = x == y
B = x == z

print("A and B:", A and B)
print("A or B:", A or B)
```

<h3>Rules of Precedence</h3>
<p>Following are the rules of precedence in the decreasing order:<br/>
() &emsp; &emsp; &emsp; Parantheses<br/>
** &emsp; &emsp; &emsp; Exponent<br/>
*, /, %, // &emsp;Multiplication, Division, Modulo, Integer Division<br/>
+, - &emsp; &emsp; &emsp; Addition, Subtraction<br/>
Operators with the same level of precedence eg + and - are evaluated as per the rules of associativity<br/>
All the operators except the exponent(**) have associativity from left to right.
</p>
